# ECSE 472 · Backward Euler transient analysis

Run the cells in order. The assignment circuit uses a 60 Hz, 1 V cosine source and a 100 µF capacitor initially at 0 V. Results start at the first completed Backward Euler step.

```python
from pathlib import Path
import subprocess
import sys

repo = Path("/content/semiconductor-portfolio")
if repo.exists():
    subprocess.run(["git", "-C", str(repo), "pull", "--ff-only"], check=True)
else:
    subprocess.run(["git", "clone", "--depth", "1", "https://github.com/Kwazy-Lemon/semiconductor-portfolio.git", str(repo)], check=True)
project = repo / "07_Fundamentals_of_Circuit_Simulation_and_Modelling" / "project_3"
```

```python
import csv
import matplotlib.pyplot as plt

csv_path = Path("/content/rectifier.csv")
subprocess.run([sys.executable, "mna_tran.py", "circuits/half_wave_rectifier.cir", "--csv", str(csv_path)], cwd=project, check=True)
with csv_path.open() as stream:
    rows = list(csv.DictReader(stream))

time_ms = [1000 * float(row["time_s"]) for row in rows]
fig, ax = plt.subplots(figsize=(9, 4.6))
for node, color in [("n1", "royalblue"), ("n2", "black")]:
    ax.plot(time_ms, [float(row[f"V({node})"]) for row in rows], color=color, label=f"V({node})")
ax.set(xlabel="Time (ms)", ylabel="Voltage (V)", title="Half-wave rectifier", xlim=(0, 50), ylim=(-1.05, 1.05))
ax.grid(alpha=0.25)
ax.legend()
fig.tight_layout()
plt.show()
```
